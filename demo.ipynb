## Failure Cases & Analysis

| Input | What happened | Why |
|---|---|---|
| `"asdkjfh qwoeiuqwe"` | Fallback message shown | Correct behavior — gibberish has no match |
| `"Do you accept bitcoin?"` | Fallback message shown | Correct behavior — genuinely not in the FAQ |
| A question that overlaps two FAQ topics, e.g. `"how do i cancel and get a refund"` | May match only one topic (e.g. cancellation) and ignore the refund part | `difflib` scores whole-string similarity, so it can't split a compound question into two answers |
| Very short inputs like `"hours"` | May fail to match even though it's related to "working hours" | `cutoff=0.6` requires fairly high similarity to the full stored question, so single-word inputs often score too low |

**Takeaway:** fuzzy string matching handles typos and light rewording well, but it has no real understanding of meaning — it can't handle compound questions or very short/keyword-style input. A production version would use semantic embeddings instead of `difflib` to fix this.
